```python
import pandas as pd
import glob

def get_best_metrics(file_path):
    # Change to pd.read_csv as available files are CSVs
    df = pd.read_csv(file_path)

    # ensure numeric
    for col in df.columns:
        df[col] = pd.to_numeric(df[col], errors='coerce')

    # pick best epoch based on mAP@0.5
    best_row = df.loc[df['mAP@0.5'].idxmax()]

    return {
        # Adjust Model name extraction for CSV files
        "Model": file_path.split("/")[-1].replace(".csv", ""),
        "Epoch": int(best_row["Epoch"]),
        "Precision": round(best_row["Precision"], 4),
        "Recall": round(best_row["Recall"], 4),
        "F1-score": round(best_row["F1-score"], 4),
        "mAP@0.5": round(best_row["mAP@0.5"], 4),
        "mAP@0.5:0.95": round(best_row["mAP@0.5:0.95"], 4),
    }

# Read all models from the available CSV files
# Use glob to find all files matching the pattern '/content/epoch_metrics_*.csv'
files = glob.glob("/content/epoch_metrics_*.csv")

results = []

for f in files:
    results.append(get_best_metrics(f))

# create final table
final_df = pd.DataFrame(results)

# sort by mAP@0.5
final_df = final_df.sort_values(by="mAP@0.5", ascending=False)

print("\nFINAL PERFORMANCE TABLE:\n")
print(final_df)

# save
final_df.to_csv("final_metrics_table.csv", index=False)
```
